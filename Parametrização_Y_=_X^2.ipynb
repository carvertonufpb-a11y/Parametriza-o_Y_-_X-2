Gera uma terna em PA na parábola y=x^2\, dado um (l, m, n) inteiros e apresenta gráficamente.

```python
# ============================================================
# Parametrização de ternas em PA na parábola y = x^2
# x_0 = p - q, x_1, x_2 = p + q
# com p^2 + q^2 = x_1^2
# ============================================================

import math

def parametrizacao_geral(l, m, n):
    """
    Recebe parâmetros inteiros l, m, n e
    retorna as diferenças r1, r2 e os pontos (x, y) na parábola.
    """
    # 1. Construção da terna pitagórica
    p = l * (n**2 - m**2)
    q = l * (2 * n * m)
    x_1 = l * (n**2 + m**2)

    # 2. Cálculo das coordenadas x_0 e x_2 (abscissas)
    x_0 = p - q
    x_2 = p + q

    # 3. Cálculo das coordenadas y (ordenadas)
    y_0 = x_0**2
    y_1 = x_1**2
    y_2 = x_2**2

    # 4. Diferenças entre as ordenadas (razões da PA)
    r1 = y_1 - y_0
    r2 = y_2 - y_1

    # 5. Organização dos pontos
    ponto_a = (x_0, y_0)
    ponto_b = (x_1, y_1)
    ponto_c = (x_2, y_2)

    return r1, r2, ponto_a, ponto_b, ponto_c


# ========================
# Programa principal
# ========================

print("Parametrização Geral de Ternas em PA na parábola y = x^2\n")

# 1. Solicitação dos Parâmetros de Entrada 'l, m e n'
while True:
    try:

        l = int(input("Digite o valor de l: "))
        m = int(input("Digite o valor de m: "))
        n = int(input("Digite o valor de n: "))


        # Validações dos parâmetros n e m para ternas pitagóricas

        if n <= m:
            print("ERRO: É necessário que n > m para uma construção válida.")
            continue # Volta para o início do loop

        if math.gcd(n, m) != 1:
            print("ERRO: n e m devem ser coprimos (mdc(n,m) = 1) para gerar ternas primitivas.")
            continue # Volta para o início do loop

        if (n - m) % 2 == 0:
            print("ERRO: n e m devem ter paridades opostas (um par e outro ímpar).")
            continue # Volta para o início do loop

        # Se todas as entradas são válidas e validações passaram, sai do loop
        break

    except ValueError:
        print("ERRO: Entrada inválida. Por favor, digite valores INTEIROS para l, m e n.")
        continue # Volta para o início do loop

# 2. Chamada da Função e Recebimento dos Resultados
r1, r2, A, B, C = parametrizacao_geral(l, m, n)

# 3. Exibição dos Resultados e Análise da PA

print("\n" + "="*50)
print(f"Parâmetros de entrada: l={l}, m={m}, n={n}")
print("="*50)

# Coordenadas dos Pontos na Parábola y = x^2 (Todos Inteiros)
print("\nI. Pontos (x, y) na Parábola:")
print(f"Ponto A: ({A[0]}, {A[1]})")
print(f"Ponto B: ({B[0]}, {B[1]})")
print(f"Ponto C: ({C[0]}, {C[1]})")

# Sequência das Ordenadas (y)
print("\nII. Sequência das Ordenadas (y) [y0, y1, y2]:")
print(f"PA: {A[1]}, {B[1]}, {C[1]}")

# Análise da Progressão Aritmética (PA) das Ordenadas
print("\nIII. Análise da Progressão Aritmética:")
print(f"Diferença 1 (y1 - y0): {r1}")
print(f"Diferença 2 (y2 - y1): {r2}")

# Confirmando a PA.
if r1 == r2:
    print(f"\n=> As ordenadas FORMAM uma Progressão Aritmética (PA) de razão: {r1}.")
else:
    print("\n=> As ordenadas NÃO FORMAM uma Progressão Aritmética (PA) com esta parametrização.")

#Gráfico

import matplotlib.pyplot as plt
import numpy as np

# Obtendo as coordenadas dos pontos A, B e C
x_coords = [A[0], B[0], C[0]]
y_coords = [A[1], B[1], C[1]]

# Calcula o intervalo dinamicamente para garantir simetria da parábola
# Pega o maior valor absoluto das coordenadas X de todos os pontos
max_abs_x = max(abs(x) for x in x_coords) if x_coords else 0

# Define o limite para o intervalo da parábola, estendendo por 5 unidades em ambos os lados
x_parabola_limit = max_abs_x + 5


# Cria um array de x para a parábola, abrangendo o limite calculado
x_parabola = np.linspace(-x_parabola_limit, x_parabola_limit, 400) # Definindo x_parabola
y_parabola = x_parabola**2

plt.figure(figsize=(8, 6))

# Plotando a parábola
plt.plot(x_parabola, y_parabola, label='y = x^2', color='blue', linestyle='solid')

# Plotando os pontos A, B, C
plt.scatter(A[0], A[1], color='red', s=100, zorder=5, label=f'Ponto A ({A[0]}, {A[1]})')
plt.scatter(B[0], B[1], color='green', s=100, zorder=5, label=f'Ponto B ({B[0]}, {B[1]})')
plt.scatter(C[0], C[1], color='purple', s=100, zorder=5, label=f'Ponto C ({C[0]}, {C[1]})')

# Adicionando rótulos e título
plt.title('Pontos A, B, C na Parábola y = x^2')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.grid(True)
plt.axhline(0, color='black',linewidth=0.5)
plt.axvline(0, color='black',linewidth=0.5)
plt.legend()
plt.gca().set_aspect('auto', adjustable='box')
plt.show()
```
